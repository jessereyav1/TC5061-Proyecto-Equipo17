### Exploración del dataset consolidado

```python
# Definir la ruta al archivo dataset_regresion_horario.parquet generado
parquet_path = Path("../data/processed/dataset_regresion_horario.parquet")

print(f"Cargando dataset de regresión desde: {parquet_path.resolve()}")

# 1. Cargar el archivo Parquet
df_regresion = pd.read_parquet(parquet_path)

# 2. Información general y tipos de datos
print("\n--- ESTRUCTURA DEL DATASET Y TIPOS DE DATOS ---")
df_regresion.info()

# 3. Resumen estadístico de la variable objetivo (demanda_viajes) y variables temporales
print("\n--- ESTADÍSTICAS DESCRIPTIVAS DE LA DEMANDA Y FEATURES ---")
display(df_regresion.describe())

# 4. Muestra de las primeras y últimas filas
print("\n--- PRIMERAS FILAS DEL DATASET ---")
display(df_regresion.head())

# 5. Tamaño real obtenido: Meses, número de viajes y estaciones
fecha_min = df_regresion['fecha_hora_inicio'].min()
fecha_max = df_regresion['fecha_hora_inicio'].max()

# Cálculo de meses únicos cubiertos en el dataset
num_meses = df_regresion['fecha_hora_inicio'].dt.to_period('M').nunique()

# Cálculo del número total de viajes (suma acumulada de la demanda)
total_viajes = df_regresion['demanda_viajes'].sum()

# Número de estaciones únicas
num_estaciones = df_regresion['origen_id'].nunique()

print(f"\n--- TAMAÑO REAL OBTENIDO DEL DATASET ---")
print(f"• Meses cubiertos (periodos mensuales): {num_meses}")
print(f"• Periodo temporal: desde {fecha_min} hasta {fecha_max}")
print(f"• Número total de viajes registrados: {total_viajes:,}")
print(f"• Número de estaciones únicas (origen_id): {num_estaciones:,}")
print(f"• Número total de registros (granularidad estación-hora): {len(df_regresion):,}")
```

### Visualización de resultados

```python
import os
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

import mlflow
import mlflow.sklearn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Importar la raíz del proyecto para ubicar las rutas correctamente
from tc5061_proyecto_equipo17.config import PROJ_ROOT

# Configurar el tracking URI hacia la base de datos SQLite del proyecto
mlflow.set_tracking_uri(f"sqlite:///{PROJ_ROOT / 'mlflow.db'}")

# 1. Cargar el dataset de características escaladas
parquet_path = PROJ_ROOT / "data" / "processed" / "features_regresion_escalados.parquet"
df = pd.read_parquet(parquet_path)

# 2. Filtrar estrictamente para el conjunto de prueba (Test: Septiembre 2026)
test_df = df[df["fecha_hora_inicio"] >= pd.to_datetime("2026-09-01")].copy()

# 3. Identificar la estación (origen_id) con mayor volumen total de viajes en test
top_station = test_df.groupby("origen_id")["demanda_viajes"].sum().idxmax()
total_viajes_top = test_df[test_df["origen_id"] == top_station]["demanda_viajes"].sum()

print(f"Estación con más viajes seleccionada: ID {top_station} ({total_viajes_top:,} viajes en Sep 2026)")

# 4. Filtrar los datos exclusivamente para esta estación ordenados cronológicamente
station_df = test_df[test_df["origen_id"] == top_station].sort_values("fecha_hora_inicio").reset_index(drop=True)

# 5. Cargar el modelo directamente usando el Run ID exacto
run_id = "362ab3b49b4c427aaac95f86b65199fc"
model_uri = f"runs:/{run_id}/model"

try:
    model = mlflow.sklearn.load_model(model_uri)
    print(f"Modelo cargado exitosamente desde MLflow (Run ID: {run_id})")
except Exception as e:
    # Plan B: Carga directa desde la ruta local de artefactos en caso de cualquier detalle con el backend
    model_path = PROJ_ROOT / "mlruns" / "2" / run_id / "artifacts" / "model"
    model = mlflow.sklearn.load_model(str(model_path))
    print(f"Modelo cargado exitosamente desde la ruta local: {model_path}")

# 6. Definir características y generar predicciones
exclude_cols = ["origen_id", "fecha_hora_inicio", "demanda_viajes"]
feature_cols = [col for col in test_df.columns if col not in exclude_cols]

X_station = station_df[feature_cols]
y_true_station = station_df["demanda_viajes"]
y_pred_station = model.predict(X_station)

# 7. Graficar la tendencia (Real vs Predicho)
plt.figure(figsize=(14, 6))
plt.plot(
    station_df["fecha_hora_inicio"], 
    y_true_station, 
    label="Demanda Real", 
    color="#1f77b4", 
    linewidth=1.8, 
    alpha=0.85
)
plt.plot(
    station_df["fecha_hora_inicio"], 
    y_pred_station, 
    label="Demanda Predicha", 
    color="#ff7f0e", 
    linestyle="--", 
    linewidth=1.8, 
    alpha=0.85
)

plt.title(f"Tendencia Horaria: Demanda Real vs Predicha - Estación Más Frecuente (ID: {top_station}) [Septiembre 2026]", fontsize=12, fontweight="bold")
plt.xlabel("Fecha y Hora", fontsize=10)
plt.ylabel("Número de Viajes", fontsize=10)
plt.legend(loc="upper right", fontsize=10)
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```
