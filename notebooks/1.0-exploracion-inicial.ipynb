```python
import pandas as pd
from pathlib import Path

# Usamos '../' para salir de la carpeta 'notebooks' y entrar a la raíz del proyecto
file_path = Path("../data/raw/datos_abiertos_2026_09.csv")

print(f"Cargando el archivo: {file_path}")

# Cargamos el archivo
df = pd.read_csv(file_path)

# 1. Nombres de columnas y tipos de datos
print("\n--- INFORMACIÓN GENERAL Y TIPOS DE DATOS ---")
df.info()

# 2. Conteo de valores nulos
print("\n--- VALORES NULOS POR COLUMNA ---")
null_counts = df.isnull().sum()
print(null_counts[null_counts > 0])

# 3. Muestra de las primeras filas
print("\n--- MUESTRA DE LAS PRIMERAS FILAS ---")
display(df.head(3))
```

### Exploración del dataset consolidado

```python
# Definir la ruta al archivo dataset_regresion_horario.parquet generado
parquet_path = Path("../data/processed/dataset_regresion_horario.parquet")

print(f"Cargando dataset de regresión desde: {parquet_path.resolve()}")

# 1. Cargar el archivo Parquet
df_regresion = pd.read_parquet(parquet_path)

# 2. Información general y tipos de datos
print("\n--- ESTRUCTURA DEL DATASET Y TIPOS DE DATOS ---")
df_regresion.info()

# 3. Resumen estadístico de la variable objetivo (demanda_viajes) y variables temporales
print("\n--- ESTADÍSTICAS DESCRIPTIVAS DE LA DEMANDA Y FEATURES ---")
display(df_regresion.describe())

# 4. Muestra de las primeras y últimas filas
print("\n--- PRIMERAS FILAS DEL DATASET ---")
display(df_regresion.head())

# 5. Tamaño real obtenido: Meses, número de viajes y estaciones
fecha_min = df_regresion['fecha_hora_inicio'].min()
fecha_max = df_regresion['fecha_hora_inicio'].max()

# Cálculo de meses únicos cubiertos en el dataset
num_meses = df_regresion['fecha_hora_inicio'].dt.to_period('M').nunique()

# Cálculo del número total de viajes (suma acumulada de la demanda)
total_viajes = df_regresion['demanda_viajes'].sum()

# Número de estaciones únicas
num_estaciones = df_regresion['origen_id'].nunique()

print(f"\n--- TAMAÑO REAL OBTENIDO DEL DATASET ---")
print(f"• Meses cubiertos (periodos mensuales): {num_meses}")
print(f"• Periodo temporal: desde {fecha_min} hasta {fecha_max}")
print(f"• Número total de viajes registrados: {total_viajes:,}")
print(f"• Número de estaciones únicas (origen_id): {num_estaciones:,}")
print(f"• Número total de registros (granularidad estación-hora): {len(df_regresion):,}")
```

### Exploración del df con features y escalado

```python
# Definir la ruta relativa al archivo features_regresion_escalados.parquet
parquet_path = Path("../data/processed/features_regresion_escalados.parquet")

print(f"Cargando dataset de características desde: {parquet_path.resolve()}")

# 1. Cargar el archivo Parquet
df_features = pd.read_parquet(parquet_path)

# --- RESPUESTA DOCUMENTADA EN EL NOTEBOOK ---
print("=" * 70)
print(" 1. VARIABLES ORIGINALES UTILIZADAS")
print("=" * 70)
print(" • 'inicio_del_viaje': Columna de marca temporal de los archivos crudos (con manejo")
print("   de formato mixto/latinoamericano para octubre 2024 e ISO para el resto).")
print(" • 'origen_id': Identificador único de la estación de origen del viaje.")

print("\n" + "=" * 70)
print(" 2. CONSTRUCCIÓN DE CADA VARIABLE OBJETIVO (Target)")
print("=" * 70)
print(" • Variable Objetivo ('demanda_viajes'):")
print("   - Se construyó agrupando el dataset crudo de 9M+ de registros a nivel horario")
print("     mediante un truncamiento por hora (.floor('h')) sobre 'inicio_del_viaje'.")
print("   - Se consolidó por cada estación ('origen_id') calculando el conteo total de")
print("     viajes iniciados (.size()) en cada hora específica.")
print(" • Variables Explicativas (Features) derivadas:")
print("   - Lags de demanda (1h, 2h, 24h, 168h) por estación.")
print("   - Estadísticas móviles (Rolling Mean/Std a 3h y 6h).")
print("   - Codificación cíclica temporal (Seno/Coseno para hora y mes).")

# 2. Información general y tipos de datos (para verificar las nuevas columnas)
print("\n" + "=" * 70)
print("--- ESTRUCTURA Y TIPOS DE DATOS DE LAS CARACTERÍSTICAS ---")
print("=" * 70)
df_features.info()

# 3. Resumen estadístico de las variables escaladas y de tiempo
print("\n--- ESTADÍSTICAS DESCRIPTIVAS ---")
display(df_features.describe())

# 4. Muestra de las primeras filas
print("\n--- MUESTRA DE LAS PRIMERAS FILAS ---")
display(df_features.head())

# 5. Listar todas las columnas disponibles para verificar los lags y rolling features
print("\n--- COLUMNAS DISPONIBLES EN EL DATASET ---")
print(list(df_features.columns))
```
